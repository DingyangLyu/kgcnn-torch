## Example: Loading and Running a Model

All models follow the same pattern: instantiate with config, pass a PyG `Batch` object, get predictions.

````python
import torch
import numpy as np
from torch_geometric.data import Data, Batch

# Create example molecular data
molecules = [
    Data(
        z=torch.tensor([6, 6, 8]),
        pos=torch.randn(3, 3),
        edge_index=torch.tensor([[0,1,1,2],[1,0,2,1]]),
        edge_weight=torch.ones(4, 1),
        y=torch.tensor([1.5]),
    ),
    Data(
        z=torch.tensor([6, 7, 6, 8]),
        pos=torch.randn(4, 3),
        edge_index=torch.tensor([[0,1,1,2,2,3],[1,0,2,1,3,2]]),
        edge_weight=torch.ones(6, 1),
        y=torch.tensor([2.3]),
    ),
]

batch = Batch.from_data_list(molecules)
print("Batch:", batch)
````

````python
from kgcnn_torch.models.gcn import GCNModel
from kgcnn_torch.models.schnet import SchNetModel
from kgcnn_torch.models.painn import PAiNNModel

# GCN -- uses atomic numbers + edge weights
gcn = GCNModel(node_dim=32, depth=3, gcn_units=32, num_targets=1)
with torch.no_grad():
    print("GCN prediction:", gcn(batch).squeeze().tolist())

# SchNet -- uses atomic numbers + positions
schnet = SchNetModel(node_dim=32, depth=3, units=64, num_targets=1)
with torch.no_grad():
    print("SchNet prediction:", schnet(batch).squeeze().tolist())

# PAiNN -- equivariant model using positions
painn = PAiNNModel(node_dim=32, depth=2, units=32, num_targets=1)
with torch.no_grad():
    print("PAiNN prediction:", painn(batch).squeeze().tolist())
````

## Training Example with trainer.fit()

The `kgcnn_torch.training.trainer` module provides a `fit()` function that handles the full training loop with validation, logging, scheduling, and early stopping.

````python
from torch_geometric.loader import DataLoader
from kgcnn_torch.training.trainer import fit
from kgcnn_torch.models.schnet import SchNetModel

# Create synthetic dataset (in real usage, load from kgcnn datasets)
np.random.seed(42)
dataset = []
for _ in range(60):
    n = np.random.randint(3, 8)
    # Random molecule-like data
    z = torch.randint(1, 10, (n,))
    pos = torch.randn(n, 3)
    # Connect each atom to its 2 nearest neighbors
    src, tgt = [], []
    for i in range(n):
        for j in range(n):
            if i != j:
                src.append(i)
                tgt.append(j)
    edge_index = torch.tensor([src, tgt])
    y = torch.tensor([float(n) * 0.5 + np.random.randn() * 0.1])
    dataset.append(Data(z=z, pos=pos, edge_index=edge_index, y=y))

train_data = dataset[:40]
val_data = dataset[40:]

train_loader = DataLoader(train_data, batch_size=16, shuffle=True)
val_loader = DataLoader(val_data, batch_size=16, shuffle=False)

# Model
model = SchNetModel(
    node_dim=32, depth=3, units=64,
    gauss_bins=20, gauss_distance=5.0,
    num_targets=1,
)

# Train
history = fit(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    optimizer=torch.optim.Adam(model.parameters(), lr=1e-3),
    loss_fn=torch.nn.MSELoss(),
    epochs=30,
    verbose=1,
)

print(f"\nFinal train loss: {history['train_loss'][-1]:.4f}")
print(f"Final val loss: {history['val_loss'][-1]:.4f}")
````

## Hyperparameter Configuration

Training can be configured via hyperparameter dictionaries. The structure mirrors the Keras `kgcnn` convention:

```python
hyper = {
    "model": {
        "class_name": "SchNetModel",
        "config": {
            "node_dim": 64,
            "depth": 6,
            "units": 128,
            "gauss_bins": 25,
            "gauss_distance": 5.0,
            "gauss_sigma": 0.4,
            "num_targets": 1,
        },
    },
    "training": {
        "optimizer": {"class_name": "Adam", "lr": 1e-3},
        "loss": "mse",
        "epochs": 500,
        "batch_size": 32,
        "early_stopping_patience": 50,
    },
    "data": {
        "dataset": "QM9Dataset",
        "data_unit": "eV",
    },
}
```
